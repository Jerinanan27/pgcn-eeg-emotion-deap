```python
import shutil, os, json, numpy as np

# Zip all ablation results
shutil.make_archive("/kaggle/working/ablation_within_results", "zip",
                    "/kaggle/working", "results_eeg")
print(f"✔ Zipped: {os.path.getsize('/kaggle/working/ablation_within_results.zip')/1e6:.1f} MB")

# Print combined summary
print("\n" + "="*60)
print("  ALL ABLATIONS — WITHIN-SUBJECT SUMMARY")
print("="*60)

ablations = {
    "Full PGCN":    "/kaggle/working/results_eeg/within_none",
    "Flat GCN":     "/kaggle/working/results_eeg/within_flat_gcn",
    "No Band Attn": "/kaggle/working/results_eeg/within_no_band_attn",
    "Struct Only":  "/kaggle/working/results_eeg/within_struct_only",
    "No Hjorth":    "/kaggle/working/results_eeg/within_no_hjorth",
}

for name, path in ablations.items():
    if not os.path.exists(path):
        print(f"\n  {name}: NOT RUN YET")
        continue
    results = {}
    for sd in sorted(os.scandir(path), key=lambda x: x.name):
        rp = f"{sd.path}/metrics.json"
        if os.path.exists(rp) and sd.name.startswith("s"):
            results[sd.name] = json.load(open(rp))
    if not results:
        print(f"\n  {name}: NO RESULTS")
        continue
    print(f"\n  {name} ({len(results)}/32 subjects)")
    for task, label in [("val_f1","Val"), ("aro_f1","Aro"),
                         ("dom_f1","Dom"), ("lik_f1","Lik"),
                         ("quad_f1","Quad")]:
        vals = [v[task] for v in results.values() if task in v]
        print(f"    {label}: {np.mean(vals):.4f} ± {np.std(vals):.4f}")
```
